1. Correlation Matrix (Movies)

```python
movies[["rating","popularity","vote_count","vote_average","budget","revenue"]].corr().round(2)
```

Movies – Overall Conclusion
Strong positive relationships exist between Budget and Revenue (0.75).
Vote Count and Revenue (0.75) also show a strong positive relationship.
Movie ratings have little influence on popularity or revenue.
Rating and Vote Average are duplicate metrics and one can be excluded from further analysis.

2. Correlation Matrix (TV Shows)

```python
tvshows[["rating","popularity","vote_count","vote_average"]].corr().round(2)
```

TV Shows – Overall Conclusion
No strong correlations were found among the TV show metrics.
The only perfect correlation is between Rating and Vote Average, indicating duplicate information.
Popularity and Vote Count have only a weak positive relationship.
Ratings are not strong predictors of popularity or audience engagement.

Final Business Insight
The Movies dataset shows meaningful business relationships, particularly between Budget, Revenue, and Vote Count, making these variables valuable for performance analysis. In contrast, the TV Shows dataset exhibits mostly weak relationships, suggesting that popularity and ratings alone are not sufficient to explain overall TV show performance.
