```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

movies = pd.read_csv("movies.csv")
tvshows = pd.read_csv("tvshows.csv")

movies["date_added"] = pd.to_datetime(movies["date_added"])
tvshows["date_added"] = pd.to_datetime(tvshows["date_added"])
```

```python
##Movies Rating Distribution
plt.figure(figsize=(8,5))
sns.histplot(movies["rating"], bins=20, kde=True)
plt.title("Movies Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
##TV Shows Rating Distribution
plt.figure(figsize=(8,5))
sns.histplot(tvshows["rating"], bins=20, kde=True)
plt.title("TV Shows Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
##Top 10 Movie Languages 
plt.figure(figsize=(10,5))

ax = movies["language"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Movie Languages")
plt.xlabel("Language")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)

for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
##Top 10 TV Show Languages
plt.figure(figsize=(10,5))

ax = tvshows["language"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 TV Show Languages")
plt.xlabel("Language")
plt.ylabel("Number of TV Shows")
plt.xticks(rotation=45)

for container in ax.containers:
    ax.bar_label(container, fmt='%d', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
##Movies Release Year Trend 
year = movies["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,5))

plt.plot(year.index, year.values, marker='o')

for x, y in zip(year.index, year.values):
    plt.text(x, y+5, str(y), ha='center', fontsize=9)

plt.title("Movies Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.grid(True)

plt.show()
```

```python
##TV Shows Release Year Trend 
year = tvshows["release_year"].value_counts().sort_index()

plt.figure(figsize=(12,5))

plt.plot(year.index, year.values, marker='o')

for x, y in zip(year.index, year.values):
    plt.text(x, y+5, str(y), ha='center', fontsize=9)

plt.title("TV Shows Released by Year")
plt.xlabel("Year")
plt.ylabel("Number of TV Shows")
plt.grid(True)

plt.show()
```

```python
##Top 10 Highest Revenue Movies
top = movies.nlargest(10, "revenue")

plt.figure(figsize=(15,6))
bars = plt.bar(top["title"], top["revenue"])

plt.title("Top 10 Highest Revenue Movies")
plt.xlabel("Movie")
plt.ylabel("Revenue")

plt.xticks(rotation=60, ha="right")

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x()+bar.get_width()/2,
             y,
             f"{y:,.0f}",
             ha="center",
             va="bottom",
             fontsize=8)

plt.tight_layout()
plt.show()
```

```python
##Top 10 Highest Budget Movies
import matplotlib.pyplot as plt

top_budget = movies.nlargest(10, "budget")

plt.figure(figsize=(10,6))
bars = plt.barh(top_budget["title"], top_budget["budget"])

for bar in bars:
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f'{bar.get_width():,.0f}',
        va='center',
        fontsize=9
    )

plt.title("Top 10 Highest Budget Movies")
plt.xlabel("Budget")
plt.ylabel("Movie")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
##Budget vs Revenue (Scatter Plot)
plt.figure(figsize=(8,6))

plt.scatter(
    movies["budget"],
    movies["revenue"],
    alpha=0.5
)

plt.title("Budget vs Revenue")
plt.xlabel("Budget")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
##Correlation Heatmap (Movies)
import matplotlib.pyplot as plt

corr = movies[
    ["rating","popularity","vote_count",
     "vote_average","budget","revenue"]
].corr()

plt.figure(figsize=(8,6))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j,
            i,
            f"{corr.iloc[i,j]:.2f}",
            ha="center",
            va="center",
            fontsize=9
        )

plt.title("Movies Correlation Heatmap")

plt.tight_layout()
plt.show()
```

```python
##Correlation Heatmap (TV Shows)
corr = tvshows[
    ["rating","popularity","vote_count","vote_average"]
].corr()

plt.figure(figsize=(6,5))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(
            j,
            i,
            f"{corr.iloc[i,j]:.2f}",
            ha="center",
            va="center",
            fontsize=9
        )

plt.title("TV Shows Correlation Heatmap")

plt.tight_layout()
plt.show()
```

Final Insights

1. Dataset Overview
The project contains 16,000 Movies and 15,991 TV Shows released between 2010 and 2025.
Movies include financial metrics such as Budget and Revenue, while TV Shows focus mainly on audience engagement metrics.
The dataset covers multiple countries, languages, and genres, making it suitable for global entertainment analysis.
    
Movies Insights
Audience Ratings
--Most movie ratings fall between 6.0 and 7.0, indicating that the majority of movies receive average to good audience feedback.
--Very few movies have ratings below 4 or perfect 10, making them rare cases.
Languages
--English dominates the movie industry with 9,534 movies.
--French, Japanese, Korean, Spanish, and Chinese are the next most common languages.
--English-language movies contribute the majority of the dataset.
Genres
--Drama is the most common movie genre.
--Comedy and Documentary follow after Drama.
Romance and Horror genres also have a significant presence.
Countries
--The United States produces the highest number of movies.
--Japan, South Korea, France, and the United Kingdom are major contributors.
--India is among the top producing countries.
Revenue
--Avengers: Endgame generated the highest revenue ($2.79 Billion).
--Other blockbuster franchises dominate the revenue rankings.
--High-budget franchise movies generally generate the highest earnings.
Budget
--Avatar: The Way of Water has the highest production budget.
--Large Hollywood productions invest hundreds of millions of dollars.
Popularity
--The Gorge is the most popular movie in the dataset.
--Recently released blockbuster movies occupy most popularity rankings.
Highest Ratings
--Several movies achieved a perfect rating of 10, but many have relatively few audience votes.
--High ratings do not always indicate commercial success.
TV Shows Insights
Audience Ratings
--Most TV show ratings range between 7 and 8, slightly higher than movies.
--Many TV shows receive consistently strong audience ratings.
Languages
--English is the most common TV show language.
--Chinese, Japanese, and Korean also contribute significantly.
Genres
--Drama is the leading TV genre.
--Reality shows rank second.
--Comedy and Documentary remain highly popular.
Countries
--The United States leads TV show production.
--China, Japan, South Korea, Canada, and the UK are major contributors.
    
Popularity
--The Late Show with Stephen Colbert is the most popular TV show.
--Talk shows and daily programs dominate popularity because they receive frequent audience engagement.
Highest Ratings
--Multiple TV shows have perfect ratings of 10, but most have relatively small vote counts.
Budget vs Revenue Analysis
--A clear positive relationship exists between Budget and Revenue.
--Movies with larger production budgets generally earn higher box office revenue.
--However, several expensive movies still underperformed, indicating that a high budget does not guarantee success.
Correlation Analysis
Movies
--Budget ↔ Revenue = 0.75 (Strong Positive Correlation)
--Bigger investments usually produce higher revenue.
--Vote Count ↔ Revenue = 0.75 (Strong Positive Correlation)
--Popular movies with more audience votes tend to generate more revenue.
----Budget ↔ Vote Count = 0.67
--High-budget movies generally attract more viewers.
--Popularity has only weak relationships with Rating and Revenue.
--Rating and Vote Average = 1.00
--These two columns contain duplicate information.
TV Shows

--No strong correlations exist among TV show variables.
--Rating and Vote Average are identical and represent duplicate information.
--Popularity has only a weak relationship with Vote Count.
--TV show success depends on factors beyond ratings and popularity alone.
    
Business Insights  
Movie Industry
--Investing in higher-budget productions generally leads to higher revenue.
--Audience engagement (Vote Count) is a strong indicator of commercial success.
--Franchise movies dominate both popularity and revenue rankings.
--Drama remains the most widely produced genre.
    
TV Show Industry
---Drama continues to dominate television content.
--English-language shows reach the largest audience.
--Talk shows and long-running series achieve exceptionally high popularity.
--Ratings alone are insufficient to measure TV show success.
    
Project Conclusion

--This project successfully analyzed 16,000 Movies and 15,991 TV Shows using Python, Pandas, Matplotlib, and Seaborn to understand trends in audience ratings, popularity, genres, languages, countries, budgets, and revenue.

Key findings include:

--Movies show a strong positive relationship between Budget and Revenue, indicating that higher production investment often results in greater financial returns.
--Vote Count is a strong indicator of commercial success, reflecting audience engagement.
--Drama is the most common genre across both Movies and TV Shows.
--English is the dominant language, while the United States is the leading content producer.
--TV Shows display weak correlations among performance metrics, suggesting that their success depends on multiple external factors beyond ratings and popularity.
--Rating and Vote Average contain identical information and one of them can be removed during feature selection.

Overall, the analysis provides valuable insights into global entertainment trends and demonstrates how Exploratory Data Analysis (EDA) can support business decision-making in the media industry.
